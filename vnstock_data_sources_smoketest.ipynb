```python
# Xuất CSV info của 100 công ty đầu tiên
from vnstock_data import Reference
import pandas as pd
import time

ref = Reference()

# Lấy danh sách symbol (equity) rồi lấy 100 mã đầu tiên
_equity = ref.equity.list()
symbols = _equity["symbol"].astype(str).tolist()[:100]

frames = []
for i, sym in enumerate(symbols, start=1):
    try:
        df = ref.company(sym).info()
        if not isinstance(df, pd.DataFrame):
            df = pd.DataFrame([{"symbol": sym, "error": "info() did not return DataFrame"}])
    except Exception as e:
        df = pd.DataFrame([{"symbol": sym, "error": str(e)}])

    # Đảm bảo có cột symbol + thứ tự request
    if "symbol" not in df.columns:
        df.insert(0, "symbol", sym)
    df["request_order"] = i

    frames.append(df)
    if i < len(symbols):
        time.sleep(0.15)  # giảm nguy cơ rate limit

merged_info = pd.concat(frames, ignore_index=True, sort=False)

out_path = "company_info_100.csv"
merged_info.to_csv(out_path, index=False, encoding="utf-8-sig")
print(f"Saved CSV to: {out_path} | rows={len(merged_info)}")
merged_info.head(5)
```

## Gợi ý cách bạn tự mở rộng test

1) Chạy `show_api(show_navigation=False)` để thấy đúng endpoint methods đang có trong phiên bản bạn cài.

2) Khi một lệnh báo `NotImplementedError` hoặc thiếu method, thường là do **source** (nguồn dữ liệu) không hỗ trợ endpoint đó.

3) Nếu muốn test dữ liệu quốc tế (experimental): `Market().crypto(...)`, `Market().forex(...)`, `Market().commodity(...)` — hãy dùng `Reference().search.symbol(...)` để tra `symbol_id`.
